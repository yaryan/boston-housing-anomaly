```python
import pandas as pd
df = pd.read_parquet('../data/processed/features.parquet')

# Group by distance-to-T bin and look at median price
df['T_DIST_BIN'] = pd.cut(df['DIST_TO_T_KM'], bins=[0, 0.4, 0.8, 1.5, 10], 
                          labels=['<400m', '400-800m', '800m-1.5km', '>1.5km'])
print("Median price by distance to T:")
print(df.groupby('T_DIST_BIN', observed=True)['LS_PRICE'].median().apply(lambda x: f'${x:,.0f}'))

# Correlation between neighborhood PSF and actual sale PSF
corr = df[['NBR_MEDIAN_PSF', 'PRICE_PER_SQFT']].corr().iloc[0,1]
print(f"\nNeighborhood PSF vs actual PSF correlation: {corr:.3f}")
```
